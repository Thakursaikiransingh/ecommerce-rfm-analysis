```python

# Step 1: reference date
ref_date = df['Order Date'].max()

# Step 2: group by customer
group = df.groupby('Customer ID')

# Step 3: calculate
recency = (ref_date - group['Order Date'].max()).dt.days
frequency = group['Order ID'].nunique()
monetary = group['Sales'].sum()

# Step 4: combine
rfm = pd.DataFrame({
    'Recency': recency,
    'Frequency': frequency,
    'Monetary': monetary
}).reset_index()

print(rfm.head())
print(rfm.shape)
```

```python
rfm['R_Score'] = pd.qcut(rfm['Recency'], 5, labels=[5,4,3,2,1])
rfm['F_Score'] = pd.qcut(rfm['Frequency'].rank(method='first'), 5, labels=[1,2,3,4,5])
rfm['M_Score'] = pd.qcut(rfm['Monetary'], 5, labels=[1,2,3,4,5])

rfm['RFM_Score'] = rfm['R_Score'].astype(str) + rfm['F_Score'].astype(str) + rfm['M_Score'].astype(str)

print(rfm.head())
```

```python
def segment_customer(row):
    score = row['RFM_Score']
    r = int(score[0])
    f = int(score[1])
    m = int(score[2])
    
    if r >= 4 and f >= 4 and m >= 4:
        return 'Champions'
    elif r >= 3 and f >= 3 and m >= 3:
        return 'Loyal Customers'
    elif r >= 4 and f <= 2:
        return 'New Customers'
    elif r >= 3 and f >= 3 and m <= 2:
        return 'Potential Loyalists'
    elif r <= 2 and f >= 3 and m >= 3:
        return 'At Risk'
    elif r == 1 and f == 1:
        return 'Lost Customers'
    else:
        return 'Need Attention'

rfm['Segment'] = rfm.apply(segment_customer, axis=1)
print(rfm['Segment'].value_counts())
```

```python
## Customer Segmentation Results

- Total customers analyzed = 793
- Need Attention = 199 (25.1%) — largest group!
- Loyal Customers = 159 (20.1%)
- Champions = 106 (13.4%) — best customers
- At Risk = 101 (12.7%) — need win back
- New Customers = 89 (11.2%)
- Lost Customers = 73 (9.2%) — already gone
- Potential Loyalists = 66 (8.3%)

### Key Finding:
- Only 13.4% are Champions — need to grow this
- 25.1% Need Attention — largest opportunity
- 101 At Risk customers need immediate action
- 73 Lost Customers need last chance offers
```
